## 課題1

各問で示されている図を出力してください。

その際、以下の要件を必ず満たすようにしてください。


1.   for文を必ず用いること
2.   変数`size`の値を変えることで自由に図形の大きさを変更できること




### 問1

以下の図形を出力してください。

```
*
**
***
****
*****
```

````python
size = 5

for i in range(1, size + 1):
    s = ""
    for j in range(i):
        s += "*"
    print(s)
````

### 問3

以下の図形を出力してください。

```
    *
   *
  *
 *
*
```

````python
size = 5

for i in range(size):
    s = ""
    for j in range(size - 1 - i):
        s += " "
    s += "*"
    print(s)
````

### 問4

以下の図形を出力してください。

```
*   *
 * *
  *
 * *
*   *
```

````python
size = 5

for i in range(size):
    s = ""
    
    # 中央からの距離を計算
    if i <= size // 2:
        spaces = i
    else:
        spaces = size - 1 - i
    
    # 左側のスペース
    for j in range(spaces):
        s += " "
    s += "*"
    
    # 中央行なら1つだけ
    if i == size // 2:
        print(s)
        continue
    
    # 中央のスペース
    middle_spaces = (size // 2 - spaces) * 2 - 1
    for j in range(middle_spaces):
        s += " "
    
    s += "*"
    
    print(s)
````

### 問5

以下の図形を出力してください。

```
    *
   ***
  *****
 *******
*********
```

````python
size = 5

for i in range(1, size + 1):
    s = ""
    
    # 左側のスペース
    for j in range(size - i):
        s += " "
    
    # *を奇数個並べる
    for j in range(i * 2 - 1):
        s += "*"
    
    print(s)
````
